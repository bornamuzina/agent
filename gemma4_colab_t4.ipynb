# Gemma 4 12B on Colab (T4)

Running `gemma4:12b` locally via Ollama on a free Colab T4.
Model layer for the autonomous agent project.

## Setup
1. Runtime → Change runtime type → **T4 GPU**
2. Run the cells top to bottom

## Notes
- Colab's image ships without `zstd`, which the Ollama
  installer needs — hence the `apt-get install zstd` cell.
- No systemd in Colab, so the server is started with
  `subprocess.Popen(["ollama", "serve"])` instead of a service.
- Use `gemma4:12b`, not `gemma4:12b-mlx` — the MLX build is
  Apple silicon only.
- Weights are 4-bit (~8 GB), fits the T4's 15 GB.
  Full precision would need ~24 GB.
- Colab wipes the disk on disconnect, so the model must be
  re-pulled each session.
